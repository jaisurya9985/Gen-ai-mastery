# 01 · Representing Text as Vectors


## 1. One-hot encoding


```python
import numpy as np

vocabulary = ["orchid", "tulip", "rose", "bicycle"]
index = {word: i for i, word in enumerate(vocabulary)}

def one_hot(word):
    if word not in index:
        raise ValueError(f"Unknown token: {word}")
    vector = np.zeros(len(vocabulary), dtype=int)
    vector[index[word]] = 1
    return vector

for word in vocabulary:
    print(f"{word:8} -> {one_hot(word)}")
```

## 2. A toy feature representation


```python
# Illustrative features, not a trained language model.
features = {
    "orchid":  [1.0, 1.0, 0.0],  # flower, grows in soil, has wheels
    "tulip":   [1.0, 1.0, 0.0],
    "rose":    [1.0, 1.0, 0.0],
    "bicycle": [0.0, 0.0, 1.0],
}
for token, vector in features.items():
    print(f"{token:8} -> {vector}")
```
