# 02 · Vector Geometry and Semantic Ranking


```python
import math
import numpy as np

def cosine_similarity(a, b):
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    if a.shape != b.shape:
        raise ValueError("Vectors must have the same shape")
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    if denom == 0:
        return 0.0
    return float(np.dot(a, b) / denom)

a = [2, 1, 0]
b = [4, 2, 0]
c = [0, 1, 3]
print("same direction:", round(cosine_similarity(a, b), 4))
print("different direction:", round(cosine_similarity(a, c), 4))
print("zero vector:", cosine_similarity([0, 0], [1, 2]))
```

## Interpreting cosine similarity


```python
documents = {
    "guide_python": ("Learn Python functions, loops, and data structures.", [0.9, 0.2, 0.1]),
    "guide_ml": ("Train machine learning models with labeled examples.", [0.7, 0.8, 0.1]),
    "garden": ("Grow herbs and vegetables in a sunny garden.", [0.1, 0.1, 0.9]),
}
query_vector = np.array([0.8, 0.7, 0.0])

ranked = sorted(
    ((cosine_similarity(query_vector, vector), key, text)
     for key, (text, vector) in documents.items()),
    reverse=True
)
for score, key, text in ranked:
    print(f"{score:.3f} | {key:12} | {text}")
```
